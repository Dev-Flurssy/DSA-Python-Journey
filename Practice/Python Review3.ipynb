```python
def create_dic_array(n):
    items = {}
    for i in range(n):
        item = input('please enter dictionary array item: ')
        items[i] = item
    return items


def insert_dic_array(a, i, value):
    values = list(a.values())

    if i < 0:
        print('Range doesnt exist')
    elif i >= len(values):
        print('Out of Range')
    else:
        values.insert(i, value)

    return {j: values[j] for j in range(len(values))}

arr = create_dic_array(5)
insert_dic_array(arr, 3, 'Pencil')
```
