```python
# Write a program to Convert a matrix to sparse Matrics

mat = [
    [0, 0, 0, 7, 0, 0, 0, 0],
    [1, 0, 0, 0, 8, 0, 0, 0],
    [2, 0, 0, 0, 0, 0, 0, 9],
    [0, 0, 0, 0, 0, 0, 5, 0],
    [0, 4, 0, 0, 0, 0, 0, 0]
]


def ToSparseMatrix(matrix):
    sparseMatrix = []
    
    for i in range(len(matrix)):
        for j in range(len(matrix[i])):
            
            if matrix[i][j] != 0:
                sparseMatrix.append([i, j, matrix[i][j]])

    return sparseMatrix

print(ToSparseMatrix(mat))
```

```python
# Write a program to delete the duplicate elements from a list


def remove_duplicate(l):
    seen = []
    for i in l:
        if i not in seen:
            seen.append(i)
    return seen 


nums = [1, 2, 2, 3, 1, 4]

print(remove_duplicate(nums))
```

```python
# Write a program  to add the following polynomials: 4x4+ 2x3 + 3x + 5  &  5x5 + 3x2 – 8

def add_polynomial(p1, p2):
    poly = {}
    exponentials = set(p1.keys()) | set(p2.keys())
    
    for expo in exponentials:
        poly[expo] = p1.get(expo, 0) + p2.get(expo, 0)
    
    return poly

p1 = {4:4, 3:2, 2:0, 1:3, 0:5} # 4x4 + 2x3 + 3x + 5
p2 = {5:5, 4:0, 3:0, 2:3, 1:0, 0:-8} # 5x5 + 3x2 – 8 

print(add_polynomial(p1, p2))
```

```python
# Write a program to remove alternate elements from a list and create a new list.

def remove_alternate(l):
    new = []
    for i in range(0, len(l), 2):
        new.append(l[i])

    return new

l = [1, 2, 3, 4, 5, 6]
print(remove_alternate(l))
```

```python
#  An array A = {3, 5, 7, 9} is given and it is full. Write a program to 
# insert a new element 11 to a list at a location without using the insert() function.

A = [3, 5, 7, 9]

def insert_location(A, index, x):
    B = []

    for i in range(len(A)):
        if i == index:
            B.append(x)
        B.append(A[i])

    if index == len(A):
        B.append(x)

    return B

print(insert_location(A, 2, 11))
```

```python
# Given a list of int egers, write a program to find the largest and smallest elements in the list without using max() and min().

def small_larg(l):
    smallest = l[0]
    largest = l[0]

    for i in l:
        if i < smallest:
            smallest = i

        if i > largest:
            largest = i

    return smallest, largest

nums = [12, 5, 8, 20, 3, 15]

small, large = small_larg(nums)

print("Smallest:", small)
print("Largest:", large)
```

```python
# Write a program to find a sub-list of a list from a given list

def sub_list(l, start, end):
    return l[start:end]

nums = [10, 20, 30, 40, 50]

print(sub_list(nums, 1, 4))
```

```python
def sub_list(l, start, end):
    sub = []

    for i in range(start, end):
        sub.append(l[i])

    return sub

nums = [10, 20, 30, 40, 50]

print(sub_list(nums, 1, 4))
```

```python
# Write a program that finds the largest number in an array.

def find_largest(arr):
    large = arr[0]

    for i in arr:
        if i > large:
            large = i
    return large

arr = [12, 45, 7, 89, 34]
find_largest(arr)    
```

```python
# Count how many even and odd numbers exist in an array.

def count_even_odd(arr):
    even = 0
    odd = 0
    
    for i in arr:
        if i % 2 == 0:
            even += 1
        else:
            odd += 1
            
    print("Even numbers:", even)
    print("Odd numbers:", odd)

arr = [10, 15, 20, 33, 42, 51]
count_even_odd(arr)
```

```python
# Display the array elements in reverse order without using reverse().

def reverse_arr(arr):
    res = []
    for i in arr:
        res.insert(0, i)

    return res

arr = [1, 2, 3, 4, 5]

print(reverse_arr(arr))
```

```python
# Calculate the average of all elements in an array

def ave(arr):
    total = 0

    for i in arr:
        total += i

    return total / len(arr)


scores = [75, 80, 90, 95, 85]
print(ave(scores))
```
