```python
# Recursion  involves breaking down a problem into smaller and smaller sub-problems until we get to a small enough problem that can be solved quickly.
# A recursive function must provide a base case that indicates when the function terminates.
# A recursive function must progress toward the base case.
# A recursive function must also call itself

def rec_fun_factorial(a):
    if a<1:
        return None
    if a == 1:
        return 1
    else:
        return rec_fun_factorial(a-1)*a
        
x = int(input('Enter a number: '))
print('The factorial of', x, 'is', rec_fun_factorial(x))
```

```python
# A decorator is a function that takes a parameter and returns another function. 
# A decorator takes a function’s result, modifies it and then returns it.


def improved_Div(func):
    def inner(a, b):
        if a < b and a == 0:
            return func(0, b)
        elif b == 0:
            return func(0, a)
        elif a < b:
            return func(b, a)
        else:
            return func(a, b)
    return inner


@improved_Div
def Div(a, b):
    return a / b


n, m = eval(input('Enter two numbers (separated by comma): '))
print('The division is', Div(n, m))
```

```python
# A @classmethod with cls is primarily used to work with class-level attributes, meaning attributes shared by all objects of the class.

class Car:
    make = "Benz"

    def __init__(self, name):
        self.name = name

    @classmethod
    def show_make(cls):
        print(cls.make)

    @classmethod
    def change_make(cls, new_make):
        cls.make = new_make

    @classmethod
    def create_car(cls):
        return cls("Speedstar")


Car.show_make()
print()

Car.change_make("Toyota")
Car.show_make()
print()

c = Car.create_car()
print(c.name)
```

```python
# __new__()  -> creates the object and can also decide if an objects get created or not

class Car:
    Make = 'Tesla'

    def __new__(cls, name, reg):
        if 'TS' not in reg.upper():
            return None

        print("Creating object...")
        return super().__new__(cls)

    def __init__(self, name, reg):
        self.name = name
        self.reg = reg

    def details(self):
        print(
            "My Car is",
            self.name,
            "with registration number",
            self.reg,
            "made by",
            self.Make
        )

    @classmethod
    def new_make(cls, new):
        cls.Make = new


car = Car("Model S", "TS1234")

if car:
    car.details()

Car.new_make("Toyota")

car2 = Car("Corolla", "TS5678")

if car2:
    car2.details()
```

```python
class Student:

    Reg_year = 2021
    Grad_year = 2025

    def __new__(cls, name, std_no, dept, age):
        if age<16:
            return None
        print("... Creating Student ...")
        return super().__new__(cls)

    def __init__(self, name, std_no, dept, age):
        self.name = name
        self.std_no = std_no
        self.dept = dept
        self.age = age

    def __str__(self):
        return (
            f"{self.name} is {self.age} years, studying {self.dept} "
            f"with student number {self.std_no} "
            f"from {self.Reg_year} to {self.Grad_year}"
        )

    @classmethod
    def change_grad_year(cls, year):
        cls.Grad_year = year


s1 = Student("Bisola", 20253807, "AI Engineering", 21)

print(s1)

Student.change_grad_year(2026)

print(s1)
```

```python
# Overriding __new__() to create an instance only once and storing that instance in a class variable (_instance). 
# Subsequent constructor calls return the already-created object instead of creating a new one.

class Database():

    _instance = None

    def __new__(cls, name):
        if cls._instance == None:
            cls._instance = super().__new__(cls)
        print("... Creating Database ...")   
        return cls._instance

    def __init__(self, name):
        self.name = name

    def display():
        print(self.name)

db1 = Database("MySQL")
print(db1.name)
print()

db2 = Database("PostgreSQL")
print(db2.name)
print()

print(db1.name)
print()

print(db1 is db2)
```

```python
class Users:

    _cache = {}

    def __new__(cls, username):
        if username in cls._cache:
            return cls._cache[username]
            
        obj = super().__new__(cls)
        cls._cache[username] = obj
        return obj

    def __init__(self, username):
        self.username = username

u1 = Users("Bisola")
u2 = Users("Bisola")

print(u1.username)
print(u1 is u2)    
```

```python
class FileWriter():

    def __init__(self, filename):
        self.filename = filename

    def __enter__(self):
        self.file = open(self.filename, 'w')
        print('File Opened...')
        return self.file

    def __exit__(self, exc_type, exc_val, exc_tb):
        print('File Closed...')
        return self.file.close()

with FileWriter('data.txt') as file:
    file.write('Testing File Creation')
```

```python
class FileManager:
    _cache = {}

    def __new__(cls, filename):
        if filename not in cls._cache:
            print("Creating object")
            cls._cache[filename] = super().__new__(cls)

        return cls._cache[filename]

    def __init__(self, filename):
        self.filename = filename

f1 = FileManager("a.txt")
f2 = FileManager("a.txt")
f3 = FileManager("b.txt")

print(f1 is f2)
print(f1 is f3)
```

```python
# __new__ → control creation and caching of objects.
# __init__ → configure the object.
# __enter__ / __exit__ → support with.
# __del__ → cleanup when garbage collected, but don't rely on it for important resources.
# For files, with is preferred over __del__ because cleanup is deterministic.
```

```python
# Static methods are used when we require a method that has nothing to do  with the instance or class variables. 
# This method can be used to interact with a different class or object.


class Employee:

    employed = True
    company = "Microsoft"
    employees = []

    def __new__(cls, work_id, name, email, role, salary):

        if not cls.employed:
            print("Employee is not employed.")
            return None

        print("... Onboarding Employee ...")
        return super().__new__(cls)

    def __init__(self, work_id, name, email, role, salary):
        self.__work_id = work_id
        self.name = name
        self.__email = email
        self.role = role
        self._salary = salary

        Employee.employees.append(self)

    @staticmethod
    def validate_email(email):
        return "@" in email

    @staticmethod
    def calculate_bonus(salary):
        return salary * 0.10

    @classmethod
    def change_company(cls, new_company):
        cls.company = new_company

    @classmethod
    def show_all_employees(cls):
        for emp in cls.employees:
            print(emp)
            print()

    def details(self):
        return (
            f'---------------------------------------------------\n'
            f'Work ID : {self.__work_id}\n'
            f'Name    : {self.name}\n'
            f'Email   : {self.__email}\n'
            f'Role    : {self.role}\n'
            f'Salary  : {self._salary}\n'
            f'Company : {self.company}\n'
            f'---------------------------------------------------'
        )

    def __str__(self):
        return  f'{self.name} - {self.role}'


emp1 = Employee(
    2002109,
    "Darasimi Williams",
    "darasimi109@mst.com",
    "Developer",
    10000
)

emp2 = Employee(
    2002110,
    "Bisola Alayande",
    "bisola@mst.com",
    "AI Engineer",
    12000
)

emp3 = Employee(
    2002111,
    "John Smith",
    "john@mst.com",
    "Data Analyst",
    9000
)

print()
print(emp1.details())
print()
print(emp2.details())
print()
print(emp3.details())

print()
Employee.show_all_employees()
```

```python
class University:

    university_name = "Oxford"

    class Student:

        def __init__(self, student_id, name):
            self.student_id = student_id
            self.name = name

        def show(self):
            print(f"Student ID: {self.student_id}")
            print(f"Name: {self.name}")

    def __init__(self, student_id, name):
        self.student = University.Student(student_id, name)

    def show(self):
        print(f"University: {University.university_name}")
        self.student.show()


u = University(101, "Bisola")
u.show()
print()

s = University.Student(201, "John")
s.show()
print()

u = University(101, "Bisola")

print(u.student.student_id)
print(u.student.name)
```
