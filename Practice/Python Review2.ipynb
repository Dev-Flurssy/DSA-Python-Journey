```python
def sum_total(*nums):
    sum = 0
    for num in (nums):
        sum += num
    return sum

print('Sum:', sum_total(23, 45))
print('Sum:', sum_total())
```

```python
# Recursion  involves breaking down a problem into smaller and smaller sub-problems until we get to a small enough problem that can be solved quickly.
# A recursive function must provide a base case that indicates when the function terminates.
# A recursive function must progress toward the base case.
# A recursive function must also call itself

def rec_fun_factorial(a):
    if a<1:
        return None
    if a == 1:
        return 1
    else:
        return rec_fun_factorial(a-1)*a
        
x = int(input('Enter a number: '))
print('The factorial of', x, 'is', rec_fun_factorial(x))
```

```python
# A lambda function is one that has no name. The name “lambda” or “anonymous” comes from the fact that they are not stated as other functions 
# using the def keyword. Rather, the lambda keyword is used to generate them and are temporary that are utilized where they are generated.

square = lambda x: x**2
print('Squared: ', square(12))
print()

maximum = lambda x, y: x if(x>y) else y
print('Maximum', maximum(10, 25))
print()

is_even = lambda x: x%2 ==0
print(is_even(8))
print(is_even(7))
print()

students = [
    ('John', 70),
    ('Mary', 90),
    ('Alex', 80)]

sorted_students = sorted(students, key=lambda student: student[1])
print(sorted_students)
```

```python
# A decorator is a function that takes a parameter and returns another function. 
# A decorator takes a function’s result, modifies it and then returns it.


def improved_Div(func):
    def inner(a, b):
        if a < b and a == 0:
            return func(0, b)
        elif b == 0:
            return func(0, a)
        elif a < b:
            return func(b, a)
        else:
            return func(a, b)
    return inner


@improved_Div
def Div(a, b):
    return a / b


n, m = eval(input('Enter two numbers (separated by comma): "))
print('The division is", Div(n, m))
```

```python
# A @classmethod with cls is primarily used to work with class-level attributes, meaning attributes shared by all objects of the class.

class Car:
    make = "Benz"

    def __init__(self, name):
        self.name = name

    @classmethod
    def show_make(cls):
        print(cls.make)

    @classmethod
    def change_make(cls, new_make):
        cls.make = new_make

    @classmethod
    def create_car(cls):
        return cls("Speedstar")


Car.show_make()
print()

Car.change_make("Toyota")
Car.show_make()
print()

c = Car.create_car()
print(c.name)
```

```python
# __new__()  -> creates the object and can also decide if an objects get created or not

class Car:
    Make = 'Tesla'

    def __new__(cls, name, reg):
        if 'TS' not in reg.upper():
            return None

        print("Creating object...")
        return super().__new__(cls)

    def __init__(self, name, reg):
        self.name = name
        self.reg = reg

    def details(self):
        print(
            "My Car is",
            self.name,
            "with registration number",
            self.reg,
            "made by",
            self.Make
        )

    @classmethod
    def new_make(cls, new):
        cls.Make = new


car = Car("Model S", "TS1234")

if car:
    car.details()

Car.new_make("Toyota")

car2 = Car("Corolla", "TS5678")

if car2:
    car2.details()
```

```python
class Student:

    Reg_year = 2021
    Grad_year = 2025

    def __new__(cls, name, std_no, dept, age):
        if age<16:
            return None
        print("... Creating Student ...")
        return super().__new__(cls)

    def __init__(self, name, std_no, dept):
        self.name = name
        self.std_no = std_no
        self.dept = dept

    def __str__(self):
        return (
            f"{self.name} is studying {self.dept} "
            f"with student number {self.std_no} "
            f"from {self.Reg_year} to {self.Grad_year}"
        )

    @classmethod
    def change_grad_year(cls, year):
        cls.Grad_year = year


s1 = Student("Bisola", 20253807, "AI Engineering")

print(s1)

Student.change_grad_year(2026)

print(s1)
```
