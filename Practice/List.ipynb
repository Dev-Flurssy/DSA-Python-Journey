```python
class Node:

    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:

    def __init__(self):
        self.head = None

    def add(self, data):

        newNode = Node(data)

        if self.head is None:
            self.head = newNode

        else:
            temp = self.head

            while temp.next:
                temp = temp.next

            temp.next = newNode

    def display(self):

        temp = self.head

        while temp:
            print(temp.data, end=' -> ')
            temp = temp.next

        print("NULL")


l1 = LinkedList()

while True:

    value = input("Enter value: ")
    l1.add(value)

    choice = input("Add more value (y/n): ")

    if choice.lower() != "y":
        break

l1.display()
```
