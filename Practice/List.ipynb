```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    # Insert at end
    def insert_end(self, data):
        newNode = Node(data)

        if self.head is None:
            self.head = newNode
            return

        temp = self.head

        while temp.next:
            temp = temp.next

        temp.next = newNode

    # Delete first node
    def del_first(self):
        if self.head is None:
            print("List is Empty")
            return

        self.head = self.head.next

    # Delete node by value
    def del_node(self, key):
        if self.head is None:
            print("List is Empty")
            return

        temp = self.head

        if temp.data == key:
            self.del_first()
            return

        temp1 = temp.next

        while temp1 is not None and temp1.data != key:
            temp = temp.next
            temp1 = temp1.next

        if temp1 is None:
            print(key, "Not Found")
        else:
            temp.next = temp1.next
            del temp1

    # Delete node after given data
    def del_after(self, data):
        if self.head is None:
            print("List is Empty")
            return

        temp = self.head

        if temp.data == data and temp.next is None:
            print("There is only one element in the list...")
            print("Next element cannot be deleted.")
            return

        while temp.data != data and temp.next is not None:
            temp = temp.next

        if temp.data != data and temp.next is None:
            print(data, "Data Not Found")

        elif temp.data == data and temp.next is None:
            print(data, "is the last node. Next node cannot be deleted.")

        else:
            temp1 = temp.next
            temp.next = temp1.next
            del temp1

    # Display list
    def display(self):
        temp = self.head

        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next

        print("None")


# Driver Program
li = LinkedList()

li.insert_end(10)
li.insert_end(20)
li.insert_end(30)
li.insert_end(40)
li.insert_end(50)

print("Original List:")
li.display()

# Delete a specific node
li.del_node(30)

print("\nAfter deleting node 30:")
li.display()

# Delete node after a given node
li.del_after(20)

print("\nAfter deleting node after 20:")
li.display()
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    # Insert node at the end
    def insert_end(self, data):
        newNode = Node(data)

        if self.head is None:
            self.head = newNode
            return

        temp = self.head

        while temp.next:
            temp = temp.next

        temp.next = newNode

    # Delete first node
    def del_first(self):
        if self.head is None:
            print("List is Empty")
            return

        self.head = self.head.next

    # Delete a node containing a specific value
    def del_node(self, key):
        if self.head is None:
            print("List is Empty")
            return

        temp = self.head

        if temp.data == key:
            self.del_first()
            return

        temp1 = temp.next

        while temp1 is not None and temp1.data != key:
            temp = temp.next
            temp1 = temp1.next

        if temp1 is None:
            print(key, "Not Found")
        else:
            temp.next = temp1.next
            del temp1

    # Delete the node after a given value
    def del_after(self, data):
        if self.head is None:
            print("List is Empty")
            return

        temp = self.head

        if temp.data == data and temp.next is None:
            print("There is only one element in the list...")
            print("Next element cannot be deleted.")
            return

        while temp.data != data and temp.next is not None:
            temp = temp.next

        if temp.data != data and temp.next is None:
            print(data, "Data Not Found")

        elif temp.data == data and temp.next is None:
            print(data, "is the last node. Next node cannot be deleted.")

        else:
            temp1 = temp.next
            temp.next = temp1.next
            del temp1

    # Display the linked list
    def display(self):
        temp = self.head

        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next

        print("None")


# Driver Program
li = LinkedList()

li.insert_end(10)
li.insert_end(20)
li.insert_end(30)
li.insert_end(40)
li.insert_end(50)

print("Original List:")
li.display()

# Delete a specific node
li.del_node(30)

print("\nAfter deleting node 30:")
li.display()

# Delete node after a given node
li.del_after(20)

print("\nAfter deleting node after 20:")
li.display()
```
