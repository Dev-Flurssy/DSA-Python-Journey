```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    
    def __init__(self):
        self.head = None

    def insert_end(self, data):
        newNode = Node(data)

        if self.head is None:
            self.head = newNode
            return

        temp = self.head

        while temp.next:
            temp = temp.next

        temp.next = newNode

    def insert_after(self, key, data):
        temp = self.head

        while temp and temp.data != key:
            temp = temp.next

        if temp is None:
            print('Not Found')
            return

        newNode = Node(data)
        newNode.next = temp.next
        temp.next = newNode

    def display(self):
        temp = self.head

        while temp:
            print(temp.data)
            temp = temp.next
        
        print('None')

li = LinkedList()

li.insert_end(10)
li.insert_end(20)
li.insert_end(30)
li.display()

li.insert_after(20, 12)
li.insert_after(10, 21)
li.insert_after(30, 39)
li.display()                
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:

    def __init__(self):
        self.head = None

    def insert_end(self, data):
        newNode = Node(data)

        if self.head is None:
            self.head = newNode
            return

        temp = self.head

        while temp.next:
            temp = temp.next

        temp.next = newNode

    def insert_before(self, key, data):

        if self.head is None:
            print("List is Empty")
            return

        # If key is in the first node
        if self.head.data == key:
            newNode = Node(data)
            newNode.next = self.head
            self.head = newNode
            return

        prev = None
        temp = self.head

        while temp and temp.data != key:
            prev = temp
            temp = temp.next

        if temp is None:
            print("Not Found")
            return

        newNode = Node(data)
        newNode.next = temp
        prev.next = newNode

    def display(self):
        temp = self.head

        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next

        print("None")


li = LinkedList()

li.insert_end(10)
li.insert_end(20)
li.insert_end(30)

print("Original List:")
li.display()

li.insert_before(20, 12)
li.insert_before(10, 21)
li.insert_before(30, 39)

print("\nAfter Insertions:")
li.display()
```
