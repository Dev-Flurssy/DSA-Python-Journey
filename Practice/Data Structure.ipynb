```python
def max_array_val(a, n):
    result = a[0]
    for i in range(n):
        if a[i] > result:
            result = a[i]

    return result
    
a = [7, 2, 15, 9]
n = 4
print(max_array_val(a, n))
```

```python
def max_array_val(a):
    result = a[0]
    for i in range(len(a)):
        if a[i] > result:
            result = a[i]

    return result
    
a = [7, 2, 15, 9]

print(max_array_val(a))
```

```python
# The Towers of Hanoi problem is an example of a recursive algorithm.
# Suppose we want to find the sum of the integers in an array.


count = 0

def sum_array(a, n):
    global count

    count += 1

    if n <= 0:
        return 0

    count += 1
    return sum_array(a, n - 1) + a[n]

a = [7, 2, 15, 9]

print(sum_array(a, len(a) - 1))
print("Count:", count)
```

```python
def sum_array(a, n):
    if n <= 0:
        return 0
    else:
        return sum_array(a, n - 1) + a[n]

a = [7, 2, 15, 9]
n = 3  # last valid index

print(sum_array(a, n))
```

```python
def sum_array(a, n):
    if n <= 0:
        return a[n]
    else:
        return sum_array(a, n - 1) + a[n]

a = [7, 2, 15, 9]
n = 3  # last valid index

print(sum_array(a, n))
```
