```python
def del_before(self, key):

    temp = self.start

    if temp.data == key:
        print("This is the first element in the list...")
        print("Previous element cannot be deleted.")
        return

    while temp.data != key and temp.next is not None:
        temp = temp.next

    if temp.data != key and temp.next is None:
        print(key, "data not found")

    elif temp.data == key and temp.prev.prev is None:
        self.del_first()

    else:
        temp1 = temp.prev

        temp1.prev.next = temp
        temp.prev = temp1.prev

        del temp1
```

```python
def del_before(self, key):

    if self.head is None or self.head.data == key:
        print("No node exists before the key")
        return

    if self.head.next and self.head.next.data == key:
        self.head = self.head.next
        return

    prev_prev = None
    prev = self.head
    curr = self.head.next

    while curr and curr.data != key:
        prev_prev = prev
        prev = curr
        curr = curr.next

    if curr is None:
        print("Key not found")
        return

    prev_prev.next = curr
```

```python
class Node:
    def __init__(self, dataval=None):
        self.data = dataval
        self.next = None


class CLinkedList:
    def __init__(self):
        self.start = Node(None)  # header node
        self.end = Node(None)    # trailer node

        self.start.next = self.end
        self.end.next = self.start
```

```python
def insert_At_Beginning(self, newdata):
    NewNode = Node(newdata)

    if self.start is None:
        self.start = NewNode
        self.end = self.start
        self.end.next = self.start
    else:
        NewNode.next = self.start
        self.start = NewNode
        self.end.next = self.start
```

```python
def insert_At_End(self, newdata):
    NewNode = Node(newdata)

    if self.start is None:
        self.insert_At_Beginning(newdata)
        return

    NewNode.next = self.start
    self.end.next = NewNode
    self.end = NewNode
```

```python
def insert_After(self, data, newdata):
    temp = self.start

    while (temp.data != data and temp.next != self.start):
        temp = temp.next

    if (temp.data != data and temp.next == self.start):
        print(data, "Not Found")

    elif (temp.data == data and temp.next == self.start):
        self.insert_At_End(newdata)

    else:
        NewNode = Node(newdata)
        NewNode.next = temp.next
        temp.next = NewNode
```

```python
def del_first(self):
    temp = self.start

    if self.start == self.end:
        self.start = self.end = None
        del temp
        return

    self.start = temp.next
    self.end.next = self.start
    del temp
```

```python
def del_last(self):
    temp = self.start

    if self.start == self.end:
        self.del_first()
        return

    while temp.next.next != self.start:
        temp = temp.next

    temp1 = temp.next
    temp.next = self.start
    self.end = temp

    del temp1
```
