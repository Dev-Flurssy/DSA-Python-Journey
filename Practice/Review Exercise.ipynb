```python
# Write a Python program that calculates compound interest.

# Compound Interest Calculator

principal = float(input("Enter principal amount: "))
rate = float(input("Enter annual interest rate (%): "))
time = float(input("Enter time in years: "))

amount = principal * (1 + rate / 100) ** time
compound_interest = amount - principal

print(f"Final Amount: {amount:.2f}")
print(f"Compound Interest: {compound_interest:.2f}")
    
```

```python
# Write a Python program that swaps two numbers using a third variable.

# Program to swap two numbers using a third variable

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("\nBefore swapping:")
print("a =", a)
print("b =", b)

# Swapping using a third variable
temp = a
a = b
b = temp

print("\nAfter swapping:")
print("a =", a)
print("b =", b)
```

```python
# Write a program that finds the greatest common divisor (GCD), or highest common factor (HCF), of two positive numbers.

# Function to find GCD of two numbers

def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

# Input
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

# Output
print("GCD =", gcd(num1, num2))
```

```python
# Write a program to remove duplicate values from the dictionary.

# Remove duplicate values from a dictionary

my_dict = {
    'a': 10,
    'b': 20,
    'c': 10,
    'd': 30,
    'e': 20
}

result = {}

for key, value in my_dict.items():
    if value not in result.values():
        result[key] = value

print("Dictionary after removing duplicates:")
print(result)
```

```python
# Write a Python program that will add square brackets to a list. If the list contains a single item [‘Peacock’] 
# and we want three more square brackets [ ], the output should be [ [ [ [‘Peacock’] ] ] ].

# Add square brackets around a list

lst = ['Peacock']
n = 3  # number of additional square brackets

for _ in range(n):
    lst = [lst]

print(lst)
```

```python
# Write a Python program to remove an empty tuple(s) from a list of tuples.

tuples_list = [(), ('Python',), (), ('AI', 'ML'), (), (1, 2)]

result = [t for t in tuples_list if t]

print("Original List:")
print(tuples_list)

print("\nList after removing empty tuples:")
print(result)
```
