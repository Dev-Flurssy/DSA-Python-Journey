```python
# Write a Python program that calculates compound interest.

principal = float(input('Please enter Principal($): '))
rate = float(input('Please enter Rate(%): '))
time = float(input('Please enter Time(years): '))

amount = principal * (1 + rate / 100) ** time
compound_interest = amount - principal 
print( f'After {time} years, an initial investment of ${principal:.2f} '
       f'at {rate}% interest grows to ${amount:.2f}. '
       f'Compound interest earned: ${compound_interest:.2f}.')
```

```python
# Write a Python program that swaps two numbers using a third variable.

def swap(a, b):
    print('Before Swap')
    print(f'a: {a}')
    print(f'b: {b}')
    print()

    c = a
    a = b
    b = c

    print('After Swap')
    print(f'a: {a}')
    print(f'b: {b}')

swap(1, 5)
```

```python
# Write a program that finds the greatest common divisor (GCD), or highest common factor (HCF), of two positive numbers.

def gcd(a, b):
    lowest = min(a, b)

    for i in range(lowest, 0, -1):
        if (a % i == 0 and b % i == 0):
            return i
        

print(gcd(9, 21))
```

```python
# Write a program to remove duplicate values from the dictionary.

def remove_dict_dup(dic):
    seen = []
    final = {}

    for key, value in dic.items():
        if value not in seen:
            seen.append(value)
            final[key] = value

    return final


foodstuffs = {
    'Protein1': 'Eggs',
    'Protein2': 'Eggs',
    'Soup1': 'Egusi',
    'Soup2': 'Egusi',
    'Diary': 'Milk',
    'Grain': 'Rice'
}

print(remove_dict_dup(foodstuffs))
```

```python
def remove_list_dup(items):
    unique = []

    for item in items:
        if item not in unique:
            unique.append(item)
            
    return unique


foods = ['Eggs', 'Milk', 'Rice', 'Eggs', 'Milk', 'Fish']

print(remove_list_dup(foods))
```

```python
# Write a Python program that will add square brackets to a list. If the list contains a single item [‘Peacock’] 
# and we want three more square brackets [ ], the output should be [ [ [ [‘Peacock’] ] ] ].

def add_brackets(item, b= 3):
    for _ in range(b):
        item = [item]
    print(item)

add_brackets(['Peacock'])
```

```python
def add_item_brackets(items):
    for _ in range(len(items)):
        items = [items]
    print(items)

add_item_brackets(['Peacock', 'Squirrel', 'Mouse'])
```

```python
# Write a Python program to remove an empty tuple(s) from a list of tuples.

def remove_emp_tuple(tups):
    result = []

    for tup in tups:
        if tup != ():
            result.append(tup)

    return tuple(result)

data = [(), ('Red',), (), ('Green', 'Purple'), ('Blue', 'Orange')]

print(remove_emp_tuple(data))
```
