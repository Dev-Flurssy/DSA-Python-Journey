```python
# Creating an array 

def create_array(n):
    nums = []
    for i in range(n):
        no = int(input('Please enter number: '))
        nums.append(no)
    return nums

print(create_array(4))
```

```python
# Access a data in the list from a given location

def access(a, i):
    if 0 <= i < len(a):
        print(f'The element at {i} is {a[i]}')
    else:
        print('The index is more than the size of array')

a = [1, 5, 9, 2, 7, 3]
n = 6
access(a, n)
```

```python
# insert a data in the list at a given location

def insert_data(a, i, x):
    if i < 0:
        print('Index is below the Range')
    elif i > len(a):
        print('Index is out of Range')
    else:
        a.insert(i, x)
        return a 

a = [1, 5, 9, 2, 7, 3]
b= 5
c = 0
insert_data(a, b, c)
```

```python
# Delete a data from the list at a given location

def del_data(a, i):
    if i < 0:
        print('Index is below the Range')
    elif i > len(a):
        print('Index is out of Range')
    else:
        del a[i]
        return a

a = [1, 5, 9, 2, 7, 3]
b = 5
del_data(a, b)       
```

```python
def create_dict_array():
    dict_A = {}
    num = int(input('Please enter number: '))
    
    for i in range(num):
        item = input('Please enter items in the Dictionary: ')
        dict_A[i] = item
    
    return dict_A

dl = create_dict_array()
print()

def insert_dict_list(dl, i, item):
    items = list(dl.values())

    if i < 0:
        print('Lower than Range')
    elif i > len(items):
        print('Out of Range')
    else:
        items.insert(i, item)

    return {j: items[j] for j in range(len(items))}

print(insert_dict_list(dl, 3, 'purple'))
```

```python
def create_dict_array():
    dict_B = {}
    num = int(input('Please enter number: '))
    
    for i in range(num):
        key = input('Please enter dict Key: ')
        value = input('Please enter dict Value: ')
        dict_B[key] = value

    return dict_B

def del_dict_array(a, i):
    items = list(a.items())
    if i < 0:
        print('Lower than Range')
    elif i >= len(items):
        print('Higher than Range')
    else:
        del items[i]
    return dict(items)

dl = create_dict_array()
del_dict_array(dl, 2)
```
