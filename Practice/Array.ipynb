```python
class Array:
    def __init__(self):
        self.maxsize = 10
        self.size = 0
        self.items = []

    def readArray(self):
        no = int(input('Please enter a number less than 10: '))

        if no <= self.maxsize:
            for i in range(no):
                data = input('Please enter item needed: ')
                self.items.append(data)
                self.size += 1

            return self.items
        else:
            print("Number of data is greater than maximum size of array")
            return self.items

    def insertArrsy(self, index, data):
        if (index > self.size) or (self.size >= self.maxsize):
            print(f'Addition of {data} is not possible at index {index}')
        else:
            self.items.append(None)
            
            for i in range(self.size, index, -1):
                self.items[i] = self.items[i - 1]
            
            self.items[index] = data
            self.size += 1

    def delArray(self, data):
        if data in self.items:
            self.items.remove(data)
            self.size -= 1
            return self.items
        
        else:
            print('This element is not in the Array!')
        return self.items
arr = Array()

arr.readArray()

arr.insertArrsy(3, 7)

print(arr.items)
print(arr.delArray(5))
```

```python
class Array:
    def __init__(self):
        self.maxsize = 10
        self.size = 0
        self.items = []

    def display(self):
        print(self.items)

    def add_ind(self, index, data):
        if index < 0 or index > self.size:
            print("Invalid index")
            return

        if self.size >= self.maxsize:
            print("Array can't take any more data")
            return

        self.items.append(None)

        for i in range(self.size, index, -1):
            self.items[i] = self.items[i - 1]

        self.items[index] = data
        self.size += 1

    def search_data(self, data):
        if data in self.items:
            for i in range(self.size):
                if self.items[i] == data:
                    print(f"{data} found at position {i}")
                    return
        print("Item not found")

    def del_ind(self, index):
        if index < 0 or index >= self.size:
            print("Element at that index doesn't exist")
        else:
            removed = self.items.pop(index)
            self.size -= 1
            print(f"Item {removed} removed")

    def del_item(self, data):
        if data in self.items:
            self.items.remove(data)
            self.size -= 1
            print(f"{data} removed")
        else:
            print("Item does not exist")


arr = Array()

# Add initial data
arr.items = ["A", "B", "C", "D"]
arr.size = len(arr.items)

print("Initial Array:")
arr.display()

arr.add_ind(2, "X")
print("\nAfter inserting X at index 2:")
arr.display()

arr.search_data("C")

arr.del_ind(3)
print("\nAfter deleting index 3:")
arr.display()

arr.del_item("B")
print("\nAfter deleting B:")
arr.display()
```

```python
def mul_poly(p1, p2):
    res = {}

    for expo1, coeff1 in p1.items():
        for expo2, coeff2 in p2.items():

            new_expo = expo1 + expo2
            new_coeff = coeff1 * coeff2

            res[new_expo] = res.get(new_expo, 0) + new_coeff

    return res
```

```python
def createMat():
    rows = int(input('Please enter Rows: '))
    cols = int(input('Please enter Cols: '))

    matrix = []

    for i in range(rows):
        row = []

        for j in range(cols):
            num = int(input(f'Please Enter element [{i}][{j}]  '))
            row.append(num)
        
        matrix.append(row)
    
    return matrix


def ToSparseMatrix(matrix):
    sparseMatrix = []

    for i in range(len(matrix)):
        for j in range(len(matrix[0])):
            if matrix[i][j] != 0:
                sparseMatrix.append([i, j, matrix[i][j]])

    return sparseMatrix


def transposeMatrix(sparseMatrix):
    for row in sparseMatrix:
        row[0], row[1] = row[1], row[0]

    sparseMatrix.sort()
    return sparseMatrix


matrix = createMat()
sparseMatrix = ToSparseMatrix(matrix)
print(sparseMatrix)
print()

transposeMatrix(sparseMatrix)
```

```python
# Add two sparse matrices

def addSparse(A, B):
    C = []

    i = 0
    j = 0

    while i < len(A) and j < len(B):

        # Same row and same column
        if A[i][0] == B[j][0] and A[i][1] == B[j][1]:

            C.append([
                A[i][0],
                A[i][1],
                A[i][2] + B[j][2]
            ])

            i += 1
            j += 1

        # A comes first
        elif (A[i][0] < B[j][0]) or \
             (A[i][0] == B[j][0] and A[i][1] < B[j][1]):

            C.append(A[i])
            i += 1

        # B comes first
        else:

            C.append(B[j])
            j += 1

    # Copy remaining elements of A
    while i < len(A):
        C.append(A[i])
        i += 1

    # Copy remaining elements of B
    while j < len(B):
        C.append(B[j])
        j += 1

    return C
```
